```python
import sklearn.datasets as sk
import pandas as pd

data = sk.fetch_california_housing()
df = pd.DataFrame(data.data,columns=data.feature_names)
df["target"] = data.target
df = df[df.columns[df.dtypes != 'object']]
df.describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sn

cor = df.corr()
cov = df.cov()
plt.figure(figsize=[20,5])
plt.subplot(1,2,1)
sn.heatmap(cor,annot=True,cmap="coolwarm")
plt.subplot(1,2,2)
sn.heatmap(cov,annot=True,cmap="coolwarm")
plt.show()
```

# multiply with 10 

```python
corxten = (df*10).corr()
covxten = (df*10).cov()
plt.figure(figsize=[20,5])
plt.subplot(1,2,1)
sn.heatmap(corxten,annot=True,cmap="coolwarm")
plt.subplot(1,2,2)
sn.heatmap(covxten,annot=True,cmap="coolwarm")
plt.show()
```
